```python
def even_odd_numer(num):
    if num%2==0:
        print("number is even")
    else:
        print("number is odd")
    
```

```python
## Call this function
even_odd_numer(24)
even_odd_numer(5)
```

```python
## function with multiple parameters
def add(a,b):
    return a+b

result=add(2,4)
print(result)
```

```python
## Default Parameters
def greet(name):
    print(f"hello {name} welcome to my world")

greet("supriya")
```

```python
### Variable Length Arguments
## Positional And Keywords arguments

def print_number(*arun):
    for number in arun:
        print(number)
```

```python
print_number(1,2,3,4,5,6,7,"arun")
```

```python
### Keywords Arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="arun",age="32",country="India")
```

```python
def print_details(*args,**kwargs):
    for val in kwargs:
        print(f"position arrgument:{val}")
        
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"arun", name="arun",age="25",country="india")
```

```python
student={"name":"arun","age":25,"garde":24}
print(student)
print(type(student))
```

```python
# Single key is slways used
student={"name":"arun","age":25,"name":24}
print(student)
```

```python
## accessing Dictionary Elements
student={"name":"arun","age":24,"grade":'A+'}
print(student)
```

```python
## Accessing Dictionary elements
print(student['grade'])
print(student['age'])
      

## Accessing using get() method
print(student.get('name'))
print(student.get('age'))
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"not available"))
```

```python
## Modifying Dicitonary Elements
## Dictionary are mutable,so you can add, update or delete elements
print(student)
```

```python
student["age"]=33 ##update value for the key
print(student)
student["address"]="india" ## added a new key and value
print(student)
student["year of pass"]=2026
print(student)
```

```python
student["garde"]='A+'
print(student)
```

```python
del student['age'] ## delete key and value pair
print(student)
```

```python
## Dictionary methods
key=student.keys()##get all the keys
print(key)

values=student.values() ##get all values
print(student)

item=student.items() ##get all key value pairs
print(item)
```

```python
## shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="arun2"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## shallow
print(student_copy1)
print(student)
```

```python
student["name"]="arun3"
print(student_copy1)
print(student)
```

```python
### Iterating Over Dictionaries
## You can use loops to iterate over dictionatries, keys,values,or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for values in student.values():
    print(values)
```

```python
## Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Disctionaries
student={
    "student1":{"name":"arun","age":32},
    "student2":{"name":"pooja","age":23}
}
print(student)
```

```python
## Access nested dictionaries elementss
print(student["student2"]["name"])
print(student["student2"]["age"])
```

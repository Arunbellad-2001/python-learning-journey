#### Example 2 stronge password checker

```python
def is_stronge_pass(password):
    if len(password)<8:
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True

print(is_stronge_pass("Arun@2001"))
print(is_stronge_pass("arun2001"))
```

##### Example 3: Calculate the Total Cost Of Items In a Shopping Cart

```python
def tcoi_cart(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['price']*item['Quanitity']

    return total_cost

def show_item(cart):
    for item in cart:
        print(item['name'],"-",item['Quanitity'])

def most_expensive(cart):
    max_item=cart[0]
    for item in cart:
        if item['price']>max_item['price']:
            max_item=item
        return max_item
            
def total_qunitity(cart):
    total=0
    for item in cart:
        total+=item['Quanitity']
    return total
        
def add_item(cart,name,price,Quanitity):
    cart.append({'name':name,'price':price,'Quanitity':Quanitity})



#### Example cart data
cart=[
    {'name':'Apple','price':25,'Quanitity':10},
    {'name':'orange','price':15,'Quanitity':10},
    {'name':'banana','price':5,'Quanitity':32},
]

## calling the function
total_cost=tcoi_cart(cart)
print(total_cost)

show_item(cart)

most_expensive(cart)

total_qunitity(cart)

add_item(cart, "Mango", 30, 5)

```

##### Example 4: Check IF a String Is Palindrome

```python
def is_palindrome(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Hello"))
```

##### Example 5: Calculate the factorials of a number using recursion

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

factorial(5)
```

##### Example 6: Validate Email Address

```python
import re

def valid_email(email):
    pattern=r'^[a-zA-Z0-9_.+-]+@[a-zA-z0-9-]+\.[a-zA-z0-9-.]+$'
    return re.match(pattern,email) is not None

print(valid_email("arunbellad20@gmail.com"))
print(valid_email("Arun2001"))
```

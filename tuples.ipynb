```python
## Packing and Unpacking tuple
## packing

paked_tuple=1,"hello",3.14
print(paked_tuple)
```

```python
a,b,c=paked_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
frist,*middle,last=numbers
print(frist)
print(middle)
print(last)
```

```python
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))

## access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
